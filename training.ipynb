```python
def mean_squared_error(y_true, y_pred):
    if len(y_true) != len(y_pred):
        raise ValueError("length mismatch")
    errors = [(a - b) ** 2 for a, b in zip(y_true, y_pred)]
    return sum(errors) / (2*len(errors))
loss = mean_squared_error([1,2, 3], [1.3, 2.3, 3.1])
print(loss)
```

```python
import numpy as np
def fit_linear_regression(X, y, alpha=0.01, steps=2000):
    m, n = X.shape
    theta = np.zeros(n)
    for _ in range(steps):
        error = X @ theta - y
        grad = (X.T @ error) / m #gradient
        theta -= alpha * grad
    return theta
# X should already include a column of ones for the bias term.
# theta is updated using all samples at once.
```
